```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('vgsales.csv')


df.head()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df_clean = df.dropna()


print("Original Dataframe shape:", {df.shape})
print("Cleaned Dataframe shape:", {df_clean.shape})

```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False).head(10)

sns.barplot(x=genre_sales.values, y=genre_sales.index, hue=genre_sales.index, palette="viridis", legend=False)

plt.title('Top 10 Video Game Genres by Global Sales (in millions)', fontsize=14, fontweight='bold')

plt.xlabel('Total Global Sales', fontsize=12)

plt.ylabel('Genre', fontsize=12)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


regions = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']
regional_totals = df[regions].sum()


plt.figure(figsize=(10, 6))
sns.barplot(x=regional_totals.index, y=regional_totals.values, palette='viridis')

plt.title('Total Video Game Sales by Region', fontsize=16)
plt.xlabel('Region', fontsize=12)
plt.ylabel('Total Sales (in millions)', fontsize=12)
plt.xticks(ticks=[0, 1, 2, 3], labels=['North America', 'Europe', 'Japan', 'Other'])
plt.show()
```

```python

yearly_regional_sales = df.groupby('Year')[regions].sum()

plt.figure(figsize=(14, 7))
sns.lineplot(data=yearly_regional_sales, dashes=False)


plt.title('Regional Video Game Sales Trends Over Time', fontsize=16)
plt.xlabel('Year of Release', fontsize=12)
plt.ylabel('Total Sales (in millions)', fontsize=12)
plt.legend(title='Region', labels=['North America', 'Europe', 'Japan', 'Other'])
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python

genre_region = df.groupby('Genre')[regions].sum()

plt.figure(figsize=(12, 8))
sns.heatmap(genre_region, annot=True, fmt='.1f', cmap='YlGnBu')
plt.title('Heatmap of Sales by Genre and Region', fontsize=16)
plt.ylabel('Genre', fontsize=12)
plt.xlabel('Region', fontsize=12)
plt.show()
```
