```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('vgsales.csv')


df.head()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df_clean = df.dropna()


print("Original Dataframe shape:", {df.shape})
print("Cleaned Dataframe shape:", {df_clean.shape})

```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False).head(10)

sns.barplot(x=genre_sales.values, y=genre_sales.index, hue=genre_sales.index, palette="viridis", legend=False)

plt.title('Top 10 Video Game Genres by Global Sales (in millions)', fontsize=14, fontweight='bold')

plt.xlabel('Total Global Sales', fontsize=12)

plt.ylabel('Genre', fontsize=12)

plt.show()
```
